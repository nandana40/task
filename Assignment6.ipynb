```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\Nandhana\Downloads\c4fe3a02-800a-48ae-b5bc-15bbc219f46a(sales_data_matplotlib).csv")
df
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly_Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.pie(df["Profit"], labels=df["Month"],autopct="%1.1f%%")
plt.title("Distribution of Total Profit by Month")
plt.show()


        
```

```python
plt.scatter(df["Customers"],df["Sales"])
plt.title("Customers vs Sales")
plt.xlabel("Customers")
plt.show()
```
