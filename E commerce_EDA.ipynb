<h1 style="font-size: 20px;">SECTION-2 PYTHON DATA CLEANING</h1>

1.CONEVRTING COLUMNS INTO APPROPRIATE DATATYPE

```python
df['Order_ID'] = df['Order_ID'].astype(str)
```

```python
df['Customer_ID'] = df['Customer_ID'].astype(str)
df['Product_ID'] = df['Product_ID'].astype(str)
df['Category'] = df['Category'].astype(str)
df['Quantity'] = df['Quantity'].astype(int)
df['Unit_Price'] = df['Unit_Price'].astype(float)
```

2.CONVERT ORDER DATE INTO A PROPER DATETIME FORMAT

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed')
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```

```python
print(df['Order_Date'].dtype)
```

3.Handle missing values appropriately.

```python
df['Discount'] = df['Discount'].fillna(df['Discount'].median())
```

```python
df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())
```

```python
categorical_columns = ['Region', 'State', 'City', 'Payment_Method']

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
categorical_columns = ['Category', 'Sub_Category', 'Region', 'State', 'City', 'Payment_Method']
```

```python
for column in categorical_columns:
    df[column] = df[column].astype(str).str.strip().str.title()
```

4.Remove or Handle Duplicate Records Appropriately.

```python
df.drop_duplicates(inplace=True)
df['Category'] = df['Category'].str.strip().str.title()
df['Sub_Category'] = df['Sub_Category'].str.strip().str.title()
df['Region'] = df['Region'].str.strip().str.title()
df['State'] = df['State'].str.strip().str.title()
df['City'] = df['City'].str.strip().str.title()
df['Payment_Method'] = df['Payment_Method'].str.strip().str.title()
df['Order_Status'] = df['Order_Status'].str.strip().str.title()
```

5.Identify and Handle Important Outliers

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df['Sales'] >= lower) & (df['Sales'] <= upper)]

Q1 = df['Profit'].quantile(0.25)
Q3 = df['Profit'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df['Profit'] >= lower) & (df['Profit'] <= upper)]
```
