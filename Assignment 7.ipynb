```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\Nandhana\Downloads\Customerr.csv")
df
```

```python
plt.plot(df["Month"],df["Sales"],marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.bar(df["Month"],df["Profit"])
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
plt.scatter(df["Marketing_Spend"],df["Sales"])
plt.title("Marketing_Spend  and Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.hist(df["Sales"],bins=6)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(df["Month"], df["Sales"], marker="o")
axes[0, 0].set_title("Monthly Sales")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True)

axes[0, 1].bar(df["Month"], df["Profit"])
axes[0, 1].set_title("Monthly Profit")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Profit")

axes[1, 0].scatter(df["Marketing_Spend"], df["Sales"])
axes[1, 0].set_title("Marketing Spend vs Sales")
axes[1, 0].set_xlabel("Marketing Spend")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].grid(True)

axes[1, 1].plot(df["Month"], df["Customers"], marker="o")
axes[1, 1].set_title("Monthly Customers")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Customers")
axes[1, 1].grid(True)


plt.show()
```
