```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\Nandhana\Downloads\numpy1\HR_Analytics_34227620.xlsx")
df
```

```python
df["Age_Category"]=df["Age"].apply(lambda x:"Young"if x <25 else "Adult"if x<=35 else "Senior")
df
```

```python
df["Travel_Status"]=df["BusinessTravel"].apply (lambda x:"Regular"if x== "Travel_Ready" else "Frequent"if x== "Travel_Frequently"else "No Travel")
df
```

```python
df["Attrition_Status"]=df["Attrition"].apply(lambda x: "Left Company"if x=="Yes" else"Working")
df
```

```python
df["Department"] = df["Department"].apply(lambda x: x.upper())
df
```

```python
df["DailyRate_Bonus"]=df["DailyRate"]+(df["DailyRate"]*0.10)
df
```

```python
df["Distance_Category"] = df["DistanceFromHome"].apply(
    lambda x: "Near" if x <= 5 else "Medium" if x <= 15 else "Far"
)
df
```

```python
df["Experience_Status"] = df["TotalWorkingYears"].apply(
    lambda x: "Experienced" if x >= 10 else "Fresher"
)
df
```
