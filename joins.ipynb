```python
#joins DataFrame using the index
df1 = pd.DataFrame({
    "Name":["Anu","Arun","Meera"]
},index=[1,2,3])

df2= pd.DataFrame({
    "Marks":[85,90,95]
},index=[1,3,2])

print(df1.join(df2))
           
```

GroupBy

```python
data = {
    "Department":["HR","IT","HR","IT","Sales"],
    "Salary":[30000,50000,35000,55000,40000]
}
df=pd.DataFrame(data)
df
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
print(df.groupby("Department")["Salary"].agg(["sum","mean","max","min","count"]))
```

applies a function to each row or column

```python
df=pd.DataFrame({
    "Marks":[80,90,95,40,39,4]})
df["Bonus"]=df["Marks"].apply(lambda x: x+5)
print(df)
```

```python
def passm(x):
    if x>=40:
        return 'pass'
    else:
        return 'fail'
```

```python
df["Result"]=df["Marks"].apply(passm)
df
```

```python
df["Result"]=df["Marks"].apply(
    lambda x: "Pass" if x>=40 else "Fail"
)
df
```

Date and Time

```python
data={'time':['2025-01-27 14:45:30','2025-02-05 09:15:00','2025-03-10 20:30:45']}
df=pd.DataFrame(data)
df
```

```python
df.dtypes
```

```python
#convert to datetime
df['time']=pd.to_datetime(df['time'])
```

```python
df.dtypes
```

```python
df['year']=df['time'].dt.year

# extract month
df['month']=df['time'].dt.month

#extract  day
df['day']=df['time'].dt.day

#xtract hour
df['hour']=df['time'].dt.hour

#extract minute
df['minute']=df['time'].dt.minute

#extract second
df['second']=df['time'].dt.second

#extract day of the week
df['dayofweek']=df['time'].dt.dayofweek

#extract week number of the year
df['week']=df['time'].dt.isocalendar().week

#extract quarter
df['quarter']=df['time'].dt.quarter

#get the name of the day(eg:monday)
df['weekday']=df['time'].dt.day_name()
```

```python
df
```

Data Transformation

Data transformation converts data into a suitable format for analysis

```python
#rename column
df.rename(columns={"Age":"Student_Age"},inplace =True)
print(df)
```

```python
df.Student_Age.dtypes
```

```python
df["Student_Age"]=df["Student_Age"].astype('int32')#float
print(df.dtypes)
```

```python
#create new column
df["Age_After_5_Years"]=df["Student_Age"] +5
print(df)
```

```python
#replace values 
df = pd.DataFrame({
    "Gender":["M","F","M"]
})
df["Gender"]=df["Gender"].replace({
    "M":"Male",
    "F":"Female"
})
print(df)
```

```python
#convert text to uppercase
df = pd.DataFrame({
    "Name":["Anu","Arun","Meera"]
})
df["Name1"]=df["Name"].str.upper()
print(df)
```

```python
#convert text to  lowercase
df["Name2"]=df["Name"].str.lower()
df
```

```python
#remove space
df = pd.DataFrame({
    "Name":[" Anu "," Arun "," Meera "]
})
df["Name"]=df["Name"].str.strip()
df
```
