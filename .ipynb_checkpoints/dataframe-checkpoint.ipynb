```python
li=[
    ['Ankit',10,'Software Developer',20000],
    ['Amal',13,'Data Analyst',50000],
    ['Anjali',15,'Data Scientist',65000],
    ['Amar',12,'Project manager',45000]
]
df=pd.DataFrame(li,columns=['EmpName','EmpID','Designation','Salary'])
df
```
